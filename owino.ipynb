# Predicting Product Returns in E-Commerce

This project uses historical order data to build a predictive model that estimates the likelihood of a product return before an order ships. By analyzing features such as product category, price, discount level, payment method, and shipping method, the model learns patterns associated with past returns. The output is a return probability for each order, allowing high-risk orders to be flagged for review before fulfillment rather than after the return has already occurred.

### 1. Business Understanding

### 1.1 Background

E-commerce has grown rapidly, allowing customers to shop online with the flexibility to return products they are not satisfied with. While this flexibility improves customer experience, it also creates significant challenges for online retailers. Product returns lead to increased reverse logistics costs, restocking efforts, lost revenue, and inefficiencies in inventory and fulfillment planning.

For many e-commerce businesses, returns are often handled reactively — only after the customer has already sent the item back. This reactive approach limits the ability to manage costs or intervene early. However, businesses generate large volumes of historical order data, including details such as product category, price, discount level, payment method, and shipping method. This data holds valuable patterns that can help predict which orders are more likely to be returned.

By analyzing historical order data, it becomes possible to identify orders at high risk of return before they are shipped. This is the foundation of the project **Predicting Product Returns in E-Commerce**, which aims to use data-driven prediction to support proactive decision-making and reduce the operational impact of product returns.

### 1.2 Problem Statement

E-commerce businesses currently have little visibility into which orders are likely to be returned until after the return has already taken place. Returns are typically identified and processed only after the customer has received the product and initiated the return request. By this stage, the business has already incurred shipping, handling, and fulfillment costs that cannot be recovered.

This reactive approach makes it difficult to reduce return-related costs or intervene early, especially when returns are concentrated in specific product categories, discount levels, or sales channels. Without a way to anticipate return risk in advance, businesses are limited to responding after losses have already occurred rather than preventing them.

This project addresses this gap by using historical order data to predict the likelihood that a given order will be returned, before it is shipped. The goal is to enable e-commerce businesses to identify high-risk orders early, allowing for proactive decisions that can reduce unnecessary returns and their associated operational costs.

### 1.3 Business Objectives

- Identify the key factors (e.g., product category, discount level, sales channel, payment method) most strongly associated with return risk.

- Demonstrate, through initial results, whether the model can realistically help catch high-risk orders before shipment.
- Deliver a basic, functional prototype (e.g., a notebook or simple Streamlit app) that shows the prediction in action, as a proof of concept rather than a full production system.

### 1.5 Business Success Criteria

This project will be considered successful from a business perspective if it meets the following criteria:

- **High-risk orders are identifiable:** The model can meaningfully distinguish orders likely to be returned from those unlikely to be returned, rather than defaulting to a single outcome.
- **Missed returns are minimized:** The model captures as many actual returns as possible, since missing a genuine return removes the opportunity for early intervention.
- **False alarms are controlled:** The model does not flag an excessive number of safe orders as high-risk, which would waste review effort and reduce trust in the system.
- **Insights are understandable:** The factors driving return risk (e.g., category, discount, channel) are clearly identifiable and explainable to non-technical stakeholders.
- **Results are delivered on time:** A working prototype and initial findings are produced within the one-month project scope, providing a foundation for future development and deployment.



### 1.6 Data Mining Goals

The business objectives above are translated into the following technical (data mining) goals:

- Build a **binary classification model** that predicts whether an order will be returned (`Returned = Yes/No`), using order-level features available before shipment.
- Estimate the **probability of return** for each order, rather than only a hard classification, to allow risk-based prioritization.
- Identify which features (e.g., product category, discount level, sales channel, payment method, shipping method) contribute most to predicting return risk.
- Evaluate the model using metrics suited to imbalanced classification, such as **precision, recall, F1-score, and ROC-AUC**, rather than accuracy alone.
- Ensure the model uses only information available at the time of order placement, avoiding data that would only be known after shipment or delivery (data leakage).
- Deliver a working prototype capable of generating return-risk predictions on new order data within the one-month project scope.



### 1.7 Data Mining Success Criteria

The technical (data mining) work will be considered successful if the following criteria are met:

- The model achieves a **recall** meaningfully better than a naive baseline (e.g., always predicting "Not Returned"), since capturing actual returns is the primary goal.
- The model achieves a reasonable **precision**, so that flagged high-risk orders are not overwhelmingly false alarms.
- The model's **ROC-AUC** and **PR-AUC** scores confirm it can distinguish returned from non-returned orders better than random guessing.
- The **confusion matrix** shows an acceptable balance between false positives and false negatives, in line with the business priorities defined in Section 1.5.
- The most influential features driving predictions (e.g., category, discount, channel) can be identified and clearly explained.
- The model pipeline — from raw data to prediction output — runs successfully end-to-end on new/unseen order data, confirming it is usable beyond the training dataset.



### 1.8 Hypotheses Guiding the Analysis

The following hypotheses will guide the exploratory analysis and modeling process. These are assumptions to be tested against the data, not confirmed facts.

- **H1:** Certain product categories (e.g., Clothing, Electronics) have a higher return rate than others, due to sizing, compatibility, or expectation-mismatch issues.
- **H2:** Orders with higher discount levels are more likely to be returned, possibly due to impulse purchases or lower perceived product value.
- **H3:** Certain sales channels (e.g., Online vs. Marketplace) are associated with different return rates, reflecting differences in customer intent or buying experience.
- **H4:** Payment method may be associated with return likelihood, as certain payment types may correlate with impulse buying or lower purchase commitment.
- **H5:** Shipping method may influence return behavior, with slower or cheaper shipping options potentially linked to different customer expectations.
- **H6:** New customers may have a higher return rate compared to returning or VIP customers, due to less familiarity with product offerings or sizing.
- **H7:** Orders with a higher quantity or unit price may carry different return risk compared to smaller, lower-value orders.

Before inspecting the structure, we first reconnect to the cloud database (Neon) to ensure the `df` DataFrame used in this section reflects the current, live dataset.

```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
import pandas as pd


# Load variables from .env file in the same folder as this notebook
load_dotenv()

PGUSER = os.getenv("PGUSER")
PGPASSWORD = os.getenv("PGPASSWORD")
PGHOST = os.getenv("PGHOST")
PGPORT = os.getenv("PGPORT")
PGDATABASE = os.getenv("PGDATABASE")

# Build the connection string
DATABASE_URL = f"postgresql+psycopg2://{PGUSER}:{PGPASSWORD}@{PGHOST}:{PGPORT}/{PGDATABASE}?sslmode=require"

engine = create_engine(DATABASE_URL)
print("Connected to cloud database (Neon) — engine ready!")
```

```python
# Load the dataset from the cloud database into a DataFrame

# SQL query to select all records from the ecommerce_orders table
query = "SELECT * FROM ecommerce_orders;"

# 2. # Run the query using the SQLAlchemy engine and load results into a Pandas DataFrame
df = pd.read_sql(query, con=engine)
```

```python
# Dataset dimensions
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```
