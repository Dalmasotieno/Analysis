```python
import psycopg2
from sqlalchemy import create_engine
import pandas as pd

# --- connection details ---
host = "localhost"      # change if your DB isn't local
port = "5432"
dbname = "ecommerce_performance_db"
user = "postgres"       # change if your PostgreSQL username is different
password = "Dalmas"

# psycopg2 connection (for direct SQL / inserts)
conn = psycopg2.connect(
    host=host, port=port, dbname=dbname, user=user, password=password
)
print("psycopg2 connected!")

# SQLAlchemy engine (best for pandas read_sql / to_sql)
engine = create_engine(f"postgresql+psycopg2://{user}:{password}@{host}:{port}/{dbname}")
print("SQLAlchemy engine ready!")
```

```python
# 1. Define your SQL query (replace 'your_table_name' with your actual table)
query = "SELECT * FROM ecommerce_orders LIMIT 100;"

# 2. Load the data directly into a Pandas DataFrame
df = pd.read_sql(query, con=engine)

# 3. Start analyzing immediately
#print(df.head())
#print(df.info())
#print(df.describe())
```
