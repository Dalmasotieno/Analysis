# Predicting Product Returns in E-Commerce

This project analyzes e-commerce order data to understand and predict which orders are likely to be returned. Using historical order details such as product category, price, discount level, sales channel, payment method, and shipping method  the goal is to build a classification model that flags high-risk return orders before they ship. 

# 1. Business Understanding

## 1.1 Business Problem

Jumia and similar e-commerce marketplaces process large volumes of customer orders daily across multiple categories, including electronics, clothing, fitness equipment, and home & office products. A portion of these orders are returned after being shipped — sometimes because of product mismatch, sizing issues, buyer's remorse, damaged items, or expectations not met.

Returns are not inherently a problem; some level of returns is normal and even expected in e-commerce. The issue arises when returns are **excessive, unpredictable, or concentrated in specific product categories, channels, or order types** without the business having early visibility into which orders are at risk.

Each return triggers a chain of costly activities: the item has to be shipped back, inspected, potentially restocked or discarded, and the original sale is effectively reversed — while the shipping, handling, and packaging costs already incurred are not recovered. If the business could identify **before an order ships** that it carries a high probability of being returned, it could make more informed decisions — such as additional verification, adjusted packaging, targeted communication with the customer, or flagging the order for review — rather than reacting only after the return has already happened.

This project focuses on using historical order data to build a model that predicts, at the point after an order is placed but before it ships, whether that order is likely to be returned. The intent is not to eliminate returns entirely, but to give the business **advance warning** on the orders most likely to result in a return, so that proactive, cost-effective decisions can be made earlier in the fulfillment process.

## 1.2 Business Context

Product returns in e-commerce carry consequences that extend well beyond the immediate lost sale. Some of the most relevant consequences for this business include:

- **Reverse logistics costs** — transporting a returned item back from the customer to a warehouse or fulfillment center involves its own shipping cost, separate from the original outbound delivery.
- **Shipping and handling costs** — the cost of the original outbound shipment is typically not recovered once an item is returned, meaning the business absorbs that cost with no corresponding revenue.
- **Inventory management problems** — returned stock must be inspected, and depending on condition, either restocked, marked down, or discarded, which complicates inventory planning and forecasting.
- **Restocking costs** — processing a returned item back into sellable inventory (inspection, repackaging, relisting) consumes staff time and warehouse resources.
- **Lost revenue** — beyond the direct cost of the return, the business loses the margin on that sale and may lose the opportunity to sell that inventory to another customer in time.
- **Customer experience** — while returns are sometimes necessary for customer satisfaction, a poor return experience (slow refunds, unclear policies) can damage trust; conversely, frequent returns tied to specific products may signal a product-quality or listing-accuracy issue worth addressing.
- **Operational inefficiencies** — high, unpredictable return volumes strain warehouse staffing, customer service capacity, and fulfillment workflows that are typically planned around expected outbound order volume, not high reverse-logistics volume.
- **Product/category-specific return patterns** — certain categories (e.g., clothing due to sizing, electronics due to compatibility or defects) may inherently carry higher return risk than others, and understanding this pattern can inform both prediction and broader business decisions (e.g., improved product descriptions, sizing guides).

## 1.4 Business Objectives

- Reduce avoidable operational costs associated with product returns.
- Improve inventory and fulfillment planning by anticipating return-related stock movement.
- Support proactive intervention (e.g., manual review, adjusted handling) before shipment rather than reactive handling after a return occurs.
- Improve organizational understanding of which factors are associated with higher return risk (e.g., category, discount, channel).



## 1.5 Data Science / Analytical Objectives

The business objectives above are translated into the following measurable analytical objectives:

- Develop a **binary classification model** that predicts whether a given order will be returned (Yes/No).
- Estimate the **probability** that a given order will be returned, rather than only a hard Yes/No label, to allow risk-based prioritization.
- Identify which order-level features (e.g., category, discount, sales channel, payment method, shipping method) are most strongly associated with return risk.
- Produce a mechanism to **flag high-risk orders before shipment**, using only information available at that point in the order lifecycle.

**Why classification is appropriate:** the business question — "will this order be returned or not?" — has a naturally binary outcome (Returned vs. Not Returned). Classification models are designed precisely to estimate the probability of a binary outcome and assign orders into risk categories, making this the appropriate technical approach rather than, for example, a regression or forecasting method.

## 1.10 Business Success Criteria

Success for this project, from a business perspective, should **not** be defined simply as "high accuracy." A model can achieve high accuracy while still failing the business if it does not correctly identify the returns that matter most, since returns are likely a minority outcome compared to non-returns.

Instead, business success should be defined in terms of:

- **Correctly identifying high-risk orders** — the model should meaningfully separate likely-to-be-returned orders from likely-to-be-kept orders, not just default to predicting the majority class.
- **Reducing false negatives where possible** — missing an order that ends up being returned means the business loses the opportunity to intervene proactively, which is central to the value of this project.
- **Controlling false positives** — flagging too many orders as high-risk when they are not could lead to unnecessary manual review effort or unwarranted customer friction, so this needs to be balanced rather than ignored.
- **Potential reduction in return-related costs** — over time, if high-risk orders are reviewed or handled differently, this should translate into fewer avoidable return-related costs (reverse logistics, restocking, lost margin).
- **Providing predictions early enough for intervention** — the prediction must be available and interpretable before the order ships; a highly accurate model that only works after shipment has no practical business value here.
- **Producing understandable insights for business users** — beyond the prediction itself, the project should surface which factors drive return risk in a way that non-technical stakeholders (e.g., category managers, operations leads) can act on.

**Why accuracy alone is insufficient:** if only, say, 10–15% of orders are returned (to be confirmed in Data Understanding), a model could achieve high accuracy simply by predicting "Not Returned" for every order — while completely failing at the actual business goal of catching risky orders. Business success must be judged on how well the model serves the *decision* it is meant to support, not on a single aggregate accuracy number.

## 1.11 Data Science Success Criteria

The following technical metrics are candidates for evaluating the model once built (not calculated in this section):

- **Precision** — of the orders flagged as high-risk, what proportion were actually returned. Relevant for controlling unnecessary interventions.
- **Recall** — of the orders that were actually returned, what proportion did the model correctly flag. Relevant for minimizing missed high-risk orders.
- **F1-score** — a balance between precision and recall, useful as a single summary metric when both error types matter.
- **ROC-AUC** — measures the model's overall ability to distinguish between returned and non-returned orders across all classification thresholds.
- **PR-AUC (Precision-Recall AUC)** — often more informative than ROC-AUC when the positive class (returns) is a minority class, which is likely the case here.
- **Confusion matrix** — provides a direct, interpretable breakdown of true positives, true negatives, false positives, and false negatives, which ties directly back to the business costs described in Section 1.12.

**Which metrics matter most and why:** given that missing a genuine return (false negative) likely carries a different business cost than incorrectly flagging a safe order (false positive), **recall and PR-AUC are likely to be prioritized** over raw accuracy, since they focus attention on how well the model captures actual returns. Precision remains important to ensure the flagging system stays practically useful and doesn't overwhelm operational teams with too many false alarms. The final weighting between precision and recall should ultimately be informed by business input on the relative cost of each error type — not decided purely by convention.

## 1.12 Cost of Prediction Errors

### False Positive
The model predicts that an order **will** be returned, but it is **not** actually returned.

**Possible business consequence:** the order may be flagged for unnecessary manual review, additional verification steps, or special handling — consuming staff time and potentially slowing down fulfillment for a customer who was never going to return the item. If this happens too frequently, it creates operational inefficiency and could even negatively affect customer experience if intervention is customer-facing.

### False Negative
The model predicts that an order **will not** be returned, but it **is** actually returned.

**Possible business consequence:** the business misses the opportunity to intervene proactively, and the order proceeds through the normal costly return cycle described in Section 1.2 (reverse logistics, restocking, lost margin) — exactly the outcome this project is meant to help reduce. This represents a missed opportunity rather than a wasted one.

**Why the two errors may carry different costs:** a false negative directly undermines the core purpose of the project (catching returns before they happen), while a false positive mainly costs some unnecessary operational effort. Depending on how costly manual review is versus how costly an unmanaged return is, the business may prefer to tolerate more false positives in exchange for fewer false negatives — but this trade-off should be confirmed with actual stakeholders rather than assumed, and no specific monetary values are assumed here since the dataset does not provide direct cost figures.

## 1.13 Prediction Timing / Data Leakage Consideration

A critical requirement for this project is that predictions must be made **before an order ships** — meaning only information that is genuinely available at or before that point in the order lifecycle can be used as a model input.

This creates an important constraint: any variable that only becomes known **after** shipment, delivery, or after a return has actually occurred **must not** be used as a feature, because doing so would allow the model to "see the future" relative to the point at which it is meant to make a prediction. This is known as **data leakage**, and it would make the model appear artificially accurate during development while being useless or misleading in real-world use, since that information simply would not exist yet at prediction time.

Examples of variables that will need to be investigated for possible leakage in the next CRISP-DM phase include (but are not limited to):
- Any field directly describing the return event itself (e.g., return reason, return date, refund amount) — these only exist because a return already happened.
- Customer satisfaction or rating fields, if they are collected only *after* delivery and could be influenced by whether the product was ultimately kept or returned.
- Any delivery-outcome or post-shipment status fields that are only populated once the order has physically reached the customer.

This project does **not** perform the actual leakage analysis or feature-by-feature leakage decisions here — that belongs in Data Understanding and Data Preparation. This section only establishes the **principle** that must guide those later decisions.
