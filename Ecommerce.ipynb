# Predicting Product Returns in E-Commerce

This project analyzes e-commerce order data to understand and predict which orders are likely to be returned. Using historical order details such as product category, price, discount level, sales channel, payment method, and shipping method  the goal is to build a classification model that flags high-risk return orders before they ship. 

# 1. Business Understanding

## 1.1 Business Problem

Jumia and similar e-commerce marketplaces process large volumes of customer orders daily across multiple categories, including electronics, clothing, fitness equipment, and home & office products. A portion of these orders are returned after being shipped — sometimes because of product mismatch, sizing issues, buyer's remorse, damaged items, or expectations not met.

Returns are not inherently a problem; some level of returns is normal and even expected in e-commerce. The issue arises when returns are **excessive, unpredictable, or concentrated in specific product categories, channels, or order types** without the business having early visibility into which orders are at risk.

Each return triggers a chain of costly activities: the item has to be shipped back, inspected, potentially restocked or discarded, and the original sale is effectively reversed — while the shipping, handling, and packaging costs already incurred are not recovered. If the business could identify **before an order ships** that it carries a high probability of being returned, it could make more informed decisions — such as additional verification, adjusted packaging, targeted communication with the customer, or flagging the order for review — rather than reacting only after the return has already happened.

This project focuses on using historical order data to build a model that predicts, at the point after an order is placed but before it ships, whether that order is likely to be returned. The intent is not to eliminate returns entirely, but to give the business **advance warning** on the orders most likely to result in a return, so that proactive, cost-effective decisions can be made earlier in the fulfillment process.
